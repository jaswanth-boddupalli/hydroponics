```python
import pandas as pd
from scipy.stats import pearsonr

#Spinach data

data = {
    'EC': [1700, 2040, 1832],
    'pH': [4.5, 4.02, 3.62],
    'TDS': [870, 1020, 970],
    'Temp': [29.2, 28.8, 29.9],
    'No. Of. Leaves': [2.1, 4.2, 5.2]
}

df = pd.DataFrame(data)
```

```python
def correlation_matrix(df):
    columns = df.columns
    n = len(columns)
    corr_matrix = pd.DataFrame(index=columns, columns=columns)

    for i in range(n):
        for j in range(n):
            corr, _ = pearsonr(df[columns[i]], df[columns[j]])
            corr_matrix.iloc[i, j] = corr

    return corr_matrix
```

```python
corr_matrix = correlation_matrix(df)
print(corr_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr_matrix = df.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
```

```python
import pandas as pd
from scipy.stats import pearsonr

#Amaranthus data

data = {
    'EC': [1730, 1980, 1802],
    'pH': [4.1, 4.5, 3.9],
    'TDS': [860, 970, 900],
    'Temp': [29.1, 30.2, 31],
    'No. Of. Leaves': [3.2, 5.2, 7.0]
}

df = pd.DataFrame(data)

def correlation_matrix(df):
    columns = df.columns
    n = len(columns)
    corr_matrix = pd.DataFrame(index=columns, columns=columns)

    for i in range(n):
        for j in range(n):
            corr, _ = pearsonr(df[columns[i]], df[columns[j]])
            corr_matrix.iloc[i, j] = corr

    return corr_matrix

corr_matrix = correlation_matrix(df)
print(corr_matrix)

import seaborn as sns
import matplotlib.pyplot as plt

corr_matrix = df.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
```
